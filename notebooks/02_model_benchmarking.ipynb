# Notebook 02: Model Benchmarking & Comparison
## AI-Based Tomato Leaf Disease Detection System

This notebook evaluates and compares all trained classical machine learning classifiers (SVM, Random Forest, AdaBoost, Gradient Boosting) against deep learning architectures (Custom CNN & EfficientNetB0).

```python
import os
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.utils.common import load_config, load_json, get_project_root
from src.evaluation.visualizer import Visualizer

cfg = load_config()
root = get_project_root()
reports_dir = root / cfg['paths']['reports_dir']
```

### 1. Load Model Comparison Results

```python
final_report_path = reports_dir / 'final_evaluation_report.json'
if final_report_path.exists():
    report = load_json(str(final_report_path))
    df_results = pd.DataFrame(report['results'])
    print(f"Best Model: {report['best_model']} (F1={report['best_f1']:.4f})")
    display(df_results)
else:
    print("Final evaluation report not found. Run main.py --phase 7 first.")
```

### 2. Plot Model Comparison Metrics

```python
if 'df_results' in locals():
    plt.figure(figsize=(10, 6))
    sns.barplot(data=df_results, x='model_name', y='accuracy', palette='magma')
    plt.title('Accuracy Comparison Across All Models', fontsize=14, fontweight='bold')
    plt.ylabel('Accuracy')
    plt.xticks(rotation=30, ha='right')
    plt.ylim(0, 1.05)
    plt.tight_layout()
    plt.show()
```
