# Notebook 01: Exploratory Data Analysis (EDA)
## AI-Based Tomato Leaf Disease Detection System

This notebook explores the PlantVillage Tomato dataset, analyzing class distributions, image dimensions, color channel statistics, and sample image visualisations.

```python
import os
import sys
from pathlib import Path

# Add project root to sys.path
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

from src.utils.common import load_config, get_project_root
from src.data.dataset import DatasetManager

cfg = load_config()
print("Project:", cfg['project']['name'])
print("Target Classes:", cfg['dataset']['class_labels'])
```

### 1. Load Dataset Statistics & Summary

```python
dm = DatasetManager(cfg)
df, train_df, val_df, test_df, stats, duplicates_df = dm.run()

print(f"Total Valid Images: {stats['total_images']}")
print(f"Train Count: {len(train_df)} | Val Count: {len(val_df)} | Test Count: {len(test_df)}")
print(f"Duplicate Count: {len(duplicates_df)}")
```

### 2. Class Distribution Visualization

```python
plt.figure(figsize=(10, 6))
sns.set_style("whitegrid")
ax = sns.barplot(x=list(stats['class_counts'].values()), y=list(stats['class_counts'].keys()), palette="viridis")
plt.title("PlantVillage Tomato Dataset — Class Distribution", fontsize=14, fontweight="bold")
plt.xlabel("Number of Images")
plt.ylabel("Disease Class")
plt.tight_layout()
plt.show()
```
